```python
import numpy as np
import json
```

```python

np.random.seed(42)


NUM_POINTS = 1000
```

```python
x = np.random.uniform(-400, 400, NUM_POINTS)
y = np.random.uniform(-400, 400, NUM_POINTS)
z = np.random.uniform(-400, 400, NUM_POINTS)

```

```python
points = []

for i in range(NUM_POINTS):
    points.append({
        "x": float(x[i]),
        "y": float(y[i]),
        "z": float(z[i])
    })
```

```python
with open("data.json", "w") as file:
    json.dump(points, file, indent=4)

print(f"{NUM_POINTS} points generated successfully!")
print("File saved as data.json")
```
