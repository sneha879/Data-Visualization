```r
barplot(
  table(data$Attrition),
  main = "Employee Attrition",
  xlab = "Attrition",
  ylab = "Number of Employees"
)
```

```r
hist(
  data$Age,
  main = "Age Distribution",
  xlab = "Age",
  ylab = "Number of Employees"
)
```

```r
boxplot(
  Age ~ Attrition,
  data = data,
  main = "Age vs Attrition",
  xlab = "Attrition",
  ylab = "Age"
)
```

```r
boxplot(
  MonthlyIncome ~ Attrition,
  data = data,
  main = "Monthly Income vs Attrition",
  xlab = "Attrition",
  ylab = "Monthly Income"
)
```
